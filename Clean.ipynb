```python

media_renda =df['renda_anual'].mean()
df.fillna({'renda_anual': media_renda}, inplace = True)
df.head(30)
```

```python
df.plot.box(column='renda_anual')
```

```python
#df['data_compra'] = pd.to_datetime(df['data_compra'], format='mixed')
df['data_compra'] = pd.to_datetime(df['data_compra'], errors='coerce')

df['data_compra_normalizada'] = df['data_compra']
df.head(30)
```

```python
df['data_compra_normalizada'] = df['data_compra_normalizada'].ffill()
df.head(30)
```

```python
df.drop(columns=['data_compra'], inplace=True)
df.head(30)
```

```python
#df_copia = df[['valor_compra', 'categoria_produto']]
valor_esporte = df[df['categoria_produto'] =='Esporte']['valor_compra'].mean()
df.loc[(df['categoria_produto'] == 'Esporte')&(df['valor_compra'].isna()), 'valor_compra'] = valor_esporte
df.head(30)
```

```python
#Finalizando o tratamento dos valores faltantes na coluna 'valor_compra' para a categoria 'Beleza'
media_valor_beleza = df[df['categoria_produto'] == 'Beleza']['valor_compra'].mean()
df.loc[(df['categoria_produto']== 'Beleza')& (df['valor_compra'].isna()), 'valor_compra'] = media_valor_beleza
```

```python
df_renda_valorcompra = df[['renda_anual', 'valor_compra']]
df_renda_valorcompra.corr()
```

```python
df_renda_normalizada =(df['renda_anual'] - df['renda_anual'].min())/ (df['renda_anual'].max()- df['renda_anual'].min())
df_renda_normalizada.head(30)
df_valor_compra_normalizado = (df['valor_compra']- df['valor_compra'].min())/(df['valor_compra'].max()- df['valor_compra'].min())
df_idade_normalizado = (df['idade']- df['idade'].min())/(df['idade'].max() - df['idade'].min())
df_normalizado = pd.concat([df_idade_normalizado, df_renda_normalizada, df_valor_compra_normalizado], axis=1)
df_normalizado.head(30)
```

```python
df['genero_encoding'] = df['genero'].astype('category').cat.codes

##df[['genero', 'genero_encoding']].head(30)
```

```python
df['genero_categorico'] = pd.Categorical(df['genero'], categories=['Masculino', 'Feminino', 'Outro'], ordered = True)
df['genero_categorico'] = df['genero_categorico'].cat.codes 
df[['genero','genero_categorico','genero_encoding']].head(30)
```
